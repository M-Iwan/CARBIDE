```python
import glob
import polars as pl
import joblib
from src.data.io import write_df
```

```python
for file in glob.glob('./data/carbide/*/*/*.joblib'):
    new_path = file.rstrip('.joblib') + '.tsv'
    df = joblib.load(file)
    df['Signature'] = df['Signature'].apply(lambda sign: ",".join(sign))
    df['DPA_confusion'] = df['DPA_confusion'].apply(lambda conf: re.sub(r'\s+', ',', str(conf).strip('[]')).lstrip(','))
    df['DPA_values'] = df['DPA_values'].apply(lambda val: re.sub(r'\s+', ',', str(val).strip('[]')).lstrip(','))
    df['DemoFP'] = df['DemoFP'].apply(lambda fp: str(fp).strip('[]').replace(' ', ','))

    write_df(df, new_path)
```
